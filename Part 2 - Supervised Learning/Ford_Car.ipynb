```python
x = df.drop('price', axis=1)
y = df['price']
```

```python
x
```

```python
y
```

```python
x.head()
```

```python
x_encoded = pd.get_dummies(x, drop_first=True)
x_encoded
```

```python
x_encoded.shape
```

```python
x_encoded = x_encoded.astype('int')
x_encoded
```

```python
from sklearn.preprocessing import LabelEncoder

# Make a copy of X
x_label = x.copy()

# Columns to encode
columns = ['model', 'transmission', 'fuelType']

# Apply LabelEncoder to each column
for col in columns:
    le = LabelEncoder()
    x_label[col] = le.fit_transform(x_label[col])
```

```python
x_label
```

```python
from sklearn.preprocessing import StandardScaler

# Identify numeric columns automatically
numeric_cols = x_encoded.select_dtypes(include=['int64', 'float64']).columns

# Initialize scaler
scaler = StandardScaler()

# Scale only numeric columns if any exist
if len(numeric_cols) > 0:
	x_encoded[numeric_cols] = scaler.fit_transform(x_encoded[numeric_cols])
```

```python
x_encoded
```

```python
x_label.columns
```

```python
x_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']] = scaler.fit_transform(x_label[['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg',
       'engineSize']])
```

```python
x_label
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_encoded, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
r2_score(y_test, y_pred)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_label, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
r2_score(y_test, y_pred)
```
